# Glacier Melt Contribution to Streamflow — Regression Analysis
### `livneh_experiments.ipynb` | Millie Spencer | 2026

This notebook implements three analyses suggested by Ben to strengthen
the glacier-streamflow attribution:

1. **Multiple Linear Regression (MLR)** — quantify how much of monthly streamflow
   variance is explained by glacier runoff vs precipitation, temperature, and PET.
   Run in two flavours:
   - **Part A**: basin-average climate from CAMELS-CL (CR2MET only, highest spatial quality)
     with all three glacier runoff estimates (CR2MET / ERA5 / CRU) as sensitivity test
   - **Part B**: glacier-location climate from OGGM lista files — fully internally
     consistent (each dataset's own glacier runoff paired with its own climate variables)

2. **Lag cross-correlation analysis** — estimate the travel time between glacier
   melt occurring on the ice surface and that water appearing at the stream gauge.
   Tests lags of 0–6 months per basin.

3. **Noah's question** — which gauged basins have ≤100 km² of glacier area,
   and do they span a useful latitudinal range?

**Key assumption**: All analyses use monthly time steps (2000–2019) to match
OGGM's simulation resolution. Streamflow from CAMELS-CL `q_m3s_mon.csv`.

## 1. Load CAMELS-CL basin-average climate data

CAMELS-CL provides basin-averaged (area-weighted) monthly climate for all 516 gauges.
We load:
- `precip_cr2met_mm_mon.csv` — basin-average precipitation (mm/month, CR2MET)
- `tmean_cr2met_C_mon.csv` — basin-average mean temperature (°C, CR2MET)
- `pet_hargreaves_mm_mon.csv` — potential evapotranspiration (mm/month, Hargreaves)

These are the highest-quality basin-level climate variables available and will be
used as the climate predictors in Part A of the regression.


```python
def load_camels_wide(fname, value_name):
    """Load a CAMELS-CL wide-format climate file, melt to long format."""
    df = pd.read_csv(f'{CAMELS_FULL}/{fname}')
    df = df[(df['year'] >= 2000) & (df['year'] <= 2019)].copy()
    id_cols = ['year', 'month']
    gauge_cols = [c for c in df.columns if c not in ['date','year','month','day']]
    df_long = df[id_cols + gauge_cols].melt(
        id_vars=id_cols, var_name='gauge_id', value_name=value_name)
    return df_long

print('Loading CAMELS-CL basin-average climate...')
precip_camels = load_camels_wide('precip_cr2met_mm_mon.csv', 'precip_mm')
tmean_camels  = load_camels_wide('tmean_cr2met_C_mon.csv',   'tmean_C')
pet_camels    = load_camels_wide('pet_hargreaves_mm_mon.csv', 'pet_mm')

# Merge all climate into one long DataFrame
camels_climate = precip_camels.merge(tmean_camels,  on=['year','month','gauge_id'])
camels_climate = camels_climate.merge(pet_camels,   on=['year','month','gauge_id'])
camels_climate['gauge_id'] = camels_climate['gauge_id'].astype(str)

print(f'CAMELS climate loaded: {len(camels_climate):,} rows')
print(f'Variables: {[c for c in camels_climate.columns if c not in ["year","month","gauge_id"]]}')
print(camels_climate.head(3))
```

## 2. Load glacier-location climate from OGGM lista files (Part B)

For the internally-consistent regression (Part B), we need basin-level climate
from the same dataset that drove each OGGM simulation. The `lista_temp.csv` and
`lista_prec.csv` files in each cluster directory contain monthly temperature (°C)
and precipitation (mm/month) at each glacier's location.

**Caveat**: these are glacier-location averages, not full basin averages.
Glacier locations are typically higher-elevation and colder than the basin centroid.
This is a known limitation flagged in the methods.

We compute the mean across all glaciers in each gauge's catchment, weighted equally.


```python
# ── Cell 2: Load glacier-location climate from OGGM lista files (Part B) ──
#
# For the internally consistent regression (Part B), we need basin-level
# climate from the same dataset that drove each OGGM simulation. The
# lista_temp.csv and lista_prec.csv files in each cluster directory contain
# monthly temperature (°C) and precipitation (mm/month) at each glacier's
# location.
#
# Caveat: glacier-location averages, not full basin averages. Glacier
# locations are typically higher-elevation and colder than the basin
# centroid. Flagged in methods.
#
# ERA5 lista files start in 1979 (not 1960 like CR2MET/CRU) because
# ERA5 reanalysis begins in 1979. Row offsets are computed per-dataset.

# Dataset-specific lista start years
LISTA_START_YEARS = {
    'CR2MET': 1960,   # 744 rows → 1960–2021
    'ERA5':   1979,   # 492 rows → 1979–2019
    'CRU':    1960,   # 744 rows → 1960–2021
}

def load_lista_for_gauge(gauge_id, ds_name, variable='temp'):
    """
    Load lista_temp or lista_prec for all glaciers in a gauge's catchment.
    Returns a DataFrame with columns: year, month, value (mean across glaciers).

    Note on lista_rgi.csv structure:
      - Row 0 is a spurious '0' header artifact — skip it
      - Rows 1..N are RGI IDs matching columns 0..N-1 of lista_temp/prec.csv
      - lista_temp.csv columns are named '0','0.1','0.2'...

    ERA5 lista files start in 1979 not 1960 — row offsets computed per dataset.
    """
    if gauge_id not in GAUGE_INFO_FRAC:
        return None
    info    = GAUGE_INFO_FRAC[gauge_id]
    rgi_ids = set(info['rgi_ids'])
    fname   = 'lista_temp.csv' if variable == 'temp' else 'lista_prec.csv'

    # Dataset-specific row offsets
    start_year = LISTA_START_YEARS.get(ds_name, 1960)
    row_start  = (2000 - start_year) * 12
    row_end    = (2020 - start_year) * 12

    rows_list = []
    for cl in info['clusters']:
        temp_path = f'{OUTPUT_ROOT}/{ds_name}/{cl}/{fname}'
        rgi_path  = f'{OUTPUT_ROOT}/{ds_name}/{cl}/lista_rgi.csv'

        if not os.path.exists(temp_path):
            continue

        df = pd.read_csv(temp_path, header=0)

        # Validate row range before slicing
        if row_start >= len(df) or row_end > len(df):
            print(f'  WARNING: {ds_name}/{cl}/{fname} has {len(df)} rows '
                  f'but need rows {row_start}:{row_end} — skipping')
            continue

        df = df.iloc[row_start:row_end]

        if os.path.exists(rgi_path):
            rgi_list = pd.read_csv(rgi_path, header=None)[0].tolist()
            rgi_list = [r for r in rgi_list if str(r).startswith('RGI')]

            if len(rgi_list) != df.shape[1]:
                # Mismatch — fall back to using all columns
                rows_list.append(df.values.mean(axis=1))
                continue

            mask = [i for i, r in enumerate(rgi_list) if r in rgi_ids]
            if mask:
                rows_list.append(df.iloc[:, mask].values.mean(axis=1))
        else:
            rows_list.append(df.values.mean(axis=1))

    if not rows_list:
        return None

    arr    = np.nanmean(np.array(rows_list), axis=0)
    n_mos  = len(arr)
    years  = np.repeat(range(2000, 2000 + n_mos // 12), 12)[:n_mos]
    months = (list(range(1, 13)) * (n_mos // 12))[:n_mos]
    return pd.DataFrame({'year': years, 'month': months, 'value': arr})

# ── Confirm row counts and start years per dataset ────────────────────────
print('Confirming lista file structure per dataset:')
for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    test_path  = f'{OUTPUT_ROOT}/{ds_name}/WA1/lista_temp.csv'
    start_year = LISTA_START_YEARS[ds_name]
    row_start  = (2000 - start_year) * 12
    row_end    = (2020 - start_year) * 12
    if os.path.exists(test_path):
        n = len(pd.read_csv(test_path, header=0))
        status = '✓' if row_end <= n else '✗ OUT OF RANGE'
        print(f'  {ds_name}: {n} rows ({n//12} yrs from {start_year})  '
              f'using rows {row_start}:{row_end}  {status}')

# ── Test on Allipen for all three datasets ────────────────────────────────
print('\nTesting load_lista_for_gauge on Allipen (9402001):')
for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    test = load_lista_for_gauge('9402001', ds_name, 'temp')
    if test is not None:
        print(f'  {ds_name}: {len(test)} rows  '
              f'sample Jan values: {test[test["month"]==1]["value"].values[:3].round(2)}')
    else:
        print(f'  {ds_name}: returned None — check paths')

# ── Build glacier-location climate cache ──────────────────────────────────
print('\nBuilding glacier-location climate cache...')
print('(~2-3 minutes for 137 gauges × 3 datasets)')

lista_climate = {}

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    n_ok = 0
    for gauge_id in GAUGE_INFO_FRAC:
        t_df = load_lista_for_gauge(gauge_id, ds_name, 'temp')
        p_df = load_lista_for_gauge(gauge_id, ds_name, 'prec')
        if t_df is not None and p_df is not None:
            merged = t_df.rename(columns={'value': 'glac_tmean_C'}).merge(
                p_df.rename(columns={'value': 'glac_prec_mm'}),
                on=['year', 'month'])
            lista_climate[(ds_name, gauge_id)] = merged
            n_ok += 1
    print(f'  {ds_name}: {n_ok} gauges loaded')

print('Done')
```

## 3. Build the monthly regression dataset

For each gauge × month × dataset combination, we assemble:

| Column | Description |
|--------|-------------|
| `q_m3s` | Observed monthly streamflow (m³/s) |
| `glacier_runoff_m3` | OGGM-simulated glacier runoff (m³/month) |
| `glac_frac` | Glacier runoff as fraction of Q |
| `precip_mm` | Basin-average precipitation from CAMELS-CL (mm/month) |
| `tmean_C` | Basin-average temperature from CAMELS-CL (°C) |
| `pet_mm` | PET from Hargreaves (mm/month) |
| `glac_tmean_C` | Glacier-location temperature from OGGM lista files |
| `glac_prec_mm` | Glacier-location precipitation from OGGM lista files |
| `month` | Calendar month (1–12, used as seasonal predictor) |
| `glacier_area_km2` | Static glacier area of basin (km²) |
| `basin_area_km2` | Total basin area (km²) |
| `lat` | Basin centroid latitude |

Only rows with complete Q and glacier runoff are kept.
Only complete years (Alexis-validated) are included.


```python
def build_regression_df(ds_name):
    """
    Build long-format monthly regression DataFrame for one climate dataset.
    Includes both CAMELS-CL basin-average climate and OGGM lista glacier-location climate.
    """
    rows = []
    for gauge_id in sorted(GAUGE_INFO_FRAC.keys()):
        res = RESULTS[ds_name].get(gauge_id)
        if res is None:
            continue
        info         = GAUGE_INFO_FRAC[gauge_id]
        m            = res['monthly']
        complete_yrs = res['complete_years']

        # Filter to complete years only
        m = m[m['year'].isin(complete_yrs)].copy()

        # CAMELS-CL basin-average climate
        gc = camels_climate[camels_climate['gauge_id'] == gauge_id]

        # OGGM lista glacier-location climate
        lc = lista_climate.get((ds_name, gauge_id))

        for _, row in m.iterrows():
            yr, mo = int(row['year']), int(row['month'])

            # Q and glacier runoff
            q_val   = row.get('q_m3s', np.nan)
            glac_m3 = row['glacier_runoff_m3']
            q_mon   = row.get('q_m3_mon', np.nan)

            if pd.isna(q_val) or q_val <= 0:
                continue

            # CAMELS-CL climate for this gauge/month
            cc = gc[(gc['year'] == yr) & (gc['month'] == mo)]
            precip = cc['precip_mm'].values[0] if len(cc) else np.nan
            tmean  = cc['tmean_C'].values[0]   if len(cc) else np.nan
            pet    = cc['pet_mm'].values[0]     if len(cc) else np.nan

            # Glacier-location climate
            glac_t = np.nan
            glac_p = np.nan
            if lc is not None:
                lc_row = lc[(lc['year'] == yr) & (lc['month'] == mo)]
                if len(lc_row):
                    glac_t = lc_row['glac_tmean_C'].values[0]
                    glac_p = lc_row['glac_prec_mm'].values[0]

            rows.append({
                'gauge_id':          gauge_id,
                'gauge_name':        info['gauge_name'],
                'year':              yr,
                'month':             mo,
                'lat':               info['cen_lat'],
                'glacier_area_km2':  info['glacier_area_km2'],
                'basin_area_km2':    info['basin_area_km2'],
                'glac_pct_basin':    info['glacier_area_km2'] / info['basin_area_km2'] * 100,
                'cluster':           info['clusters'][0],
                'dataset':           ds_name,
                # Target
                'q_m3s':             q_val,
                'q_m3_mon':          q_mon,
                # Glacier predictor
                'glacier_runoff_m3': glac_m3,
                'glac_frac':         glac_m3 / q_mon if (not pd.isna(q_mon) and q_mon > 0) else np.nan,
                # CAMELS-CL basin-average climate
                'precip_mm':         precip,
                'tmean_C':           tmean,
                'pet_mm':            pet,
                # OGGM lista glacier-location climate
                'glac_tmean_C':      glac_t,
                'glac_prec_mm':      glac_p,
                # Seasonal
                'sin_month':         np.sin(2 * np.pi * mo / 12),
                'cos_month':         np.cos(2 * np.pi * mo / 12),
            })

    df = pd.DataFrame(rows)
    print(f'{ds_name}: {len(df):,} monthly observations, '
          f'{df["gauge_id"].nunique()} gauges')
    return df

print('Building regression datasets...')
REG_DATA = {}
for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    REG_DATA[ds_name] = build_regression_df(ds_name)

print('\nDone. Sample (CR2MET):')
print(REG_DATA['CR2MET'][['gauge_id','year','month','q_m3s',
                           'glacier_runoff_m3','precip_mm',
                           'tmean_C','pet_mm']].head(5))
```

## 4. Part A — Panel regression using CAMELS-CL basin-average climate

**Motivation** (from collaborator discussion): A multiple linear regression lets us
quantify the *relative importance* of glacier melt vs precipitation vs temperature
in driving monthly streamflow. The coefficient on glacier runoff (β_glacier) tells
us how much Q increases per unit increase in glacier melt, holding all other
predictors constant. Comparing this coefficient across CR2MET / ERA5 / CRU tells
us how sensitive the result is to climate forcing choice.

**Part A setup:**
- Climate predictors: CAMELS-CL basin-average CR2MET precip, tmean, PET (same for all 3 regressions)
- Glacier predictor: changes across CR2MET / ERA5 / CRU
- Fixed effects: seasonal cycle encoded as sin/cos of month
- All predictors standardised (z-scored) so coefficients are comparable in magnitude
- Run as: (i) pooled panel across all 137 basins, (ii) per-basin separately

**Interpretation**: A larger standardised β_glacier = glacier melt is a stronger
independent driver of Q variation. R² tells you total variance explained.


```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score

PART_A_PREDICTORS = [
    'glacier_runoff_m3',   # key glacier predictor
    'precip_mm',           # CAMELS-CL basin precip (CR2MET)
    'tmean_C',             # CAMELS-CL basin temp (CR2MET)
    'pet_mm',              # PET Hargreaves
    'sin_month',           # seasonal cycle
    'cos_month',
]
PRED_LABELS = {
    'glacier_runoff_m3': 'Glacier runoff',
    'precip_mm':         'Precipitation',
    'tmean_C':           'Temperature',
    'pet_mm':            'PET',
    'sin_month':         'Season (sin)',
    'cos_month':         'Season (cos)',
}

def run_panel_regression(df, predictors, target='q_m3s', label=''):
    """
    Standardised OLS panel regression across all basins.
    Returns dict with coefficients, R2, and n.
    """
    sub = df[predictors + [target]].dropna()
    if len(sub) < 30:
        return None

    X = sub[predictors].values
    y = sub[target].values

    scaler = StandardScaler()
    X_std  = scaler.fit_transform(X)

    reg    = LinearRegression().fit(X_std, y)
    y_pred = reg.predict(X_std)
    r2     = r2_score(y, y_pred)

    # Bootstrap confidence intervals on coefficients (1000 samples)
    n_boot = 1000
    boot_coefs = np.zeros((n_boot, len(predictors)))
    rng = np.random.default_rng(42)
    for i in range(n_boot):
        idx = rng.integers(0, len(sub), len(sub))
        reg_b = LinearRegression().fit(X_std[idx], y[idx])
        boot_coefs[i] = reg_b.coef_

    ci_lo = np.percentile(boot_coefs, 2.5,  axis=0)
    ci_hi = np.percentile(boot_coefs, 97.5, axis=0)

    return {
        'label':      label,
        'n':          len(sub),
        'r2':         r2,
        'coef':       dict(zip(predictors, reg.coef_)),
        'ci_lo':      dict(zip(predictors, ci_lo)),
        'ci_hi':      dict(zip(predictors, ci_hi)),
        'intercept':  reg.intercept_,
        'scaler':     scaler,
    }

print('Running Part A panel regressions (CAMELS-CL climate)...')
PANEL_RESULTS_A = {}

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    result = run_panel_regression(
        REG_DATA[ds_name], PART_A_PREDICTORS,
        label=f'{ds_name} (panel, CAMELS-CL climate)')
    PANEL_RESULTS_A[ds_name] = result
    if result:
        print(f'\n{ds_name}:  R² = {result["r2"]:.3f}  n = {result["n"]:,}')
        for pred in PART_A_PREDICTORS:
            lo = result["ci_lo"][pred]
            hi = result["ci_hi"][pred]
            sig = '*' if (lo > 0 or hi < 0) else ''
            print(f'  {PRED_LABELS.get(pred, pred):20s}: '
                  f'{result["coef"][pred]:+.4f}  '
                  f'[{lo:+.4f}, {hi:+.4f}] {sig}')
```

### 4b. Per-basin regressions (Part A)

Running the regression separately for each basin reveals which basins show the
strongest glacier melt signal vs precipitation signal. Basins where β_glacier is
large and significant are the ones where glacier contributions are most detectable
in the streamflow record — regardless of their absolute glacier fraction.


```python
# ── Cell 4b: Per-basin regressions (Part A) ───────────────────────────────
print('Running per-basin regressions (Part A)...')
PER_BASIN_A = {}

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    basin_results = []
    df = REG_DATA[ds_name]

    for gauge_id, grp in df.groupby('gauge_id'):
        result = run_panel_regression(
            grp, PART_A_PREDICTORS, label=gauge_id)
        if result is None:
            continue
        info = GAUGE_INFO_FRAC.get(gauge_id, {})
        basin_results.append({
            'gauge_id':         gauge_id,
            'gauge_name':       info.get('gauge_name', '')[:35],
            'lat':              info.get('cen_lat', np.nan),
            'cluster':          info.get('clusters', ['?'])[0],
            'glacier_area_km2': info.get('glacier_area_km2', np.nan),
            'n_obs':            result['n'],
            'r2':               result['r2'],
            **{f'coef_{p}': result['coef'][p] for p in PART_A_PREDICTORS},
            **{f'ci_lo_{p}': result['ci_lo'][p] for p in PART_A_PREDICTORS},
            **{f'ci_hi_{p}': result['ci_hi'][p] for p in PART_A_PREDICTORS},
        })

    if not basin_results:
        print(f'  {ds_name}: 0 basins fitted — check RESULTS and REG_DATA')
        PER_BASIN_A[ds_name] = pd.DataFrame()
        continue

    PER_BASIN_A[ds_name] = pd.DataFrame(basin_results).sort_values(
        'lat', ascending=False)
    print(f'  {ds_name}: {len(PER_BASIN_A[ds_name])} basins fitted')

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    if not PER_BASIN_A[ds_name].empty:
        PER_BASIN_A[ds_name].to_csv(
            f'{REG_DIR}/per_basin_regression_partA_{ds_name}.csv', index=False)
print('Per-basin results saved')
```

## 5. Part B — Internally consistent regression using glacier-location climate

In Part B, each dataset uses its own glacier runoff **and** its own temperature
and precipitation (from the OGGM `lista_temp.csv` / `lista_prec.csv` files).
This gives fully internally consistent regressions at the cost of glacier-location
rather than basin-average climate.

**Key difference from Part A**: precipitation and temperature now vary across
CR2MET / ERA5 / CRU, not just glacier runoff. This means any change in β_glacier
between datasets could reflect both the different glacier runoff estimates AND the
different climate context — so results should be compared with Part A to disentangle.


```python
# ── Cell 5: Part B — Internally consistent regression using glacier-location climate ──
#
# Each dataset uses its own glacier runoff AND its own temperature and
# precipitation from the OGGM lista_temp/lista_prec files. This gives
# fully internally consistent regressions at the cost of glacier-location
# rather than basin-average climate.
#
# Key difference from Part A: precipitation and temperature now vary across
# CR2MET / ERA5 / CRU, not just glacier runoff. Any change in β_glacier
# between datasets could reflect both different glacier runoff estimates
# AND different climate context — compare with Part A to disentangle.

PART_B_PREDICTORS = [
    'glacier_runoff_m3',  # key glacier predictor
    'glac_prec_mm',       # glacier-location precip (dataset-specific)
    'glac_tmean_C',       # glacier-location temp (dataset-specific)
    'pet_mm',             # PET Hargreaves (same, from CAMELS-CL)
    'sin_month',
    'cos_month',
]
PRED_LABELS_B = {
    'glacier_runoff_m3': 'Glacier runoff',
    'glac_prec_mm':      'Precip (glac loc)',
    'glac_tmean_C':      'Temp (glac loc)',
    'pet_mm':            'PET',
    'sin_month':         'Season (sin)',
    'cos_month':         'Season (cos)',
}

# ── Panel regression Part B ───────────────────────────────────────────────
print('Running Part B panel regressions (glacier-location climate)...')
PANEL_RESULTS_B = {}

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    result = run_panel_regression(
        REG_DATA[ds_name], PART_B_PREDICTORS,
        label=f'{ds_name} (panel, glacier-location climate)')
    PANEL_RESULTS_B[ds_name] = result
    if result:
        print(f'\n{ds_name}:  R² = {result["r2"]:.3f}  n = {result["n"]:,}')
        for pred in PART_B_PREDICTORS:
            lo  = result['ci_lo'][pred]
            hi  = result['ci_hi'][pred]
            sig = '*' if (lo > 0 or hi < 0) else ''
            print(f'  {PRED_LABELS_B.get(pred, pred):22s}: '
                  f'{result["coef"][pred]:+.4f}  '
                  f'[{lo:+.4f}, {hi:+.4f}] {sig}')
    else:
        print(f'\n{ds_name}: regression returned None — '
              f'check lista_climate coverage for this dataset')

# ── Per-basin Part B ──────────────────────────────────────────────────────
print('\nRunning per-basin regressions (Part B)...')
PER_BASIN_B = {}

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    basin_results = []
    df = REG_DATA[ds_name]

    for gauge_id, grp in df.groupby('gauge_id'):
        result = run_panel_regression(grp, PART_B_PREDICTORS, label=gauge_id)
        if result is None:
            continue
        info = GAUGE_INFO_FRAC.get(gauge_id, {})
        basin_results.append({
            'gauge_id':         gauge_id,
            'gauge_name':       info.get('gauge_name', '')[:35],
            'lat':              info.get('cen_lat', np.nan),
            'cluster':          info.get('clusters', ['?'])[0],
            'glacier_area_km2': info.get('glacier_area_km2', np.nan),
            'n_obs':            result['n'],
            'r2':               result['r2'],
            **{f'coef_{p}': result['coef'][p] for p in PART_B_PREDICTORS},
            **{f'ci_lo_{p}': result['ci_lo'][p] for p in PART_B_PREDICTORS},
            **{f'ci_hi_{p}': result['ci_hi'][p] for p in PART_B_PREDICTORS},
        })

    if not basin_results:
        print(f'  {ds_name}: 0 basins fitted — lista_climate likely '
              f'incomplete for this dataset')
        PER_BASIN_B[ds_name] = pd.DataFrame()
        continue

    PER_BASIN_B[ds_name] = pd.DataFrame(basin_results).sort_values(
        'lat', ascending=False)
    PER_BASIN_B[ds_name].to_csv(
        f'{REG_DIR}/per_basin_regression_partB_{ds_name}.csv', index=False)
    print(f'  {ds_name}: {len(PER_BASIN_B[ds_name])} basins fitted')

# ── Diagnose lista_climate coverage ──────────────────────────────────────
print('\nDiagnosing lista_climate coverage:')
for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    count = sum(1 for k in lista_climate if k[0] == ds_name)
    n_obs = len(REG_DATA[ds_name].dropna(subset=PART_B_PREDICTORS))
    print(f'  {ds_name}: {count} gauges in lista_climate  '
          f'| {n_obs:,} complete obs for Part B regression')
```

```python
# ── Cell 5b: Exploring per-basin glacier coefficients (Part A) ────────────
#
# Answers four key questions about the per-basin regression results:
# 1. Which basins have the highest β_glacier?
# 2. Which basins show a statistically significant glacier signal?
# 3. Does β_glacier correlate with glacier area or fraction?
# 4. Do the three datasets agree on which basins are significant?

pred     = 'coef_glacier_runoff_m3'
lo_col   = 'ci_lo_glacier_runoff_m3'
hi_col   = 'ci_hi_glacier_runoff_m3'
DS_LIST  = ['CR2MET', 'ERA5', 'CRU']

# ── Question 1: Top 15 basins by β_glacier (CR2MET) ──────────────────────
print('=' * 65)
print('Q1: Top 15 basins by β_glacier (CR2MET, Part A)')
print('=' * 65)
df_cr2 = PER_BASIN_A['CR2MET'].copy()
df_cr2['sig'] = (df_cr2[lo_col] > 0) | (df_cr2[hi_col] < 0)
top15  = df_cr2.nlargest(15, pred)
print(f'{"gauge_id":10s} {"name":35s} {"lat":6s} {"β_glac":8s} '
      f'{"CI_lo":8s} {"CI_hi":8s} {"R²":6s} {"sig"}')
print('-' * 90)
for _, row in top15.iterrows():
    print(f'{row["gauge_id"]:10s} {row["gauge_name"]:35s} '
          f'{row["lat"]:6.2f} {row[pred]:8.2f} '
          f'{row[lo_col]:8.2f} {row[hi_col]:8.2f} '
          f'{row["r2"]:6.3f} {"*" if row["sig"] else ""}')

# ── Question 2: Significant vs non-significant basins ────────────────────
print('\n' + '=' * 65)
print('Q2: Significance summary across datasets')
print('=' * 65)
for ds_name in DS_LIST:
    df = PER_BASIN_A[ds_name].copy()
    df['sig'] = (df[lo_col] > 0) | (df[hi_col] < 0)
    n_sig   = df['sig'].sum()
    n_total = len(df)
    med_r2  = df['r2'].median()
    print(f'\n{ds_name}: {n_sig}/{n_total} basins significant '
          f'({n_sig/n_total*100:.0f}%)  |  median R² = {med_r2:.3f}')
    print(f'  Significant basins by cluster:')
    for cl, grp in df[df['sig']].groupby('cluster'):
        print(f'    {cl}: {len(grp)} basins')

# ── Question 3: β_glacier vs glacier area and fraction ───────────────────
print('\n' + '=' * 65)
print('Q3: Correlation of β_glacier with glacier characteristics')
print('=' * 65)

# Need mean_ann_frac from ANN_SUMMARIES if available
for ds_name in DS_LIST:
    df = PER_BASIN_A[ds_name].copy()
    df['sig'] = (df[lo_col] > 0) | (df[hi_col] < 0)

    r_area, p_area = stats.pearsonr(
        np.log10(df['glacier_area_km2'].clip(lower=0.01)),
        df[pred])
    print(f'\n{ds_name}:')
    print(f'  β_glacier vs log10(glacier area):  r={r_area:.3f}  p={p_area:.3e}')

    # Among significant basins only
    sig = df[df['sig']]
    if len(sig) > 5:
        r_sig, p_sig = stats.pearsonr(
            np.log10(sig['glacier_area_km2'].clip(lower=0.01)),
            sig[pred])
        print(f'  (significant basins only, n={len(sig)}): '
              f'r={r_sig:.3f}  p={p_sig:.3e}')

# ── Question 4: Cross-dataset agreement on significance ───────────────────
print('\n' + '=' * 65)
print('Q4: Cross-dataset agreement on significant basins')
print('=' * 65)

sig_sets = {}
for ds_name in DS_LIST:
    df = PER_BASIN_A[ds_name].copy()
    df['sig'] = (df[lo_col] > 0) | (df[hi_col] < 0)
    sig_sets[ds_name] = set(df[df['sig']]['gauge_id'])

all_gauges   = set(PER_BASIN_A['CR2MET']['gauge_id'])
agree_all    = sig_sets['CR2MET'] & sig_sets['ERA5'] & sig_sets['CRU']
agree_none   = (all_gauges - sig_sets['CR2MET'] -
                sig_sets['ERA5'] - sig_sets['CRU'])
disagree     = all_gauges - agree_all - agree_none

print(f'  Significant in ALL 3 datasets:    {len(agree_all)} basins  '
      f'({len(agree_all)/len(all_gauges)*100:.0f}%) ← robust signal')
print(f'  Non-significant in ALL 3:         {len(agree_none)} basins  '
      f'({len(agree_none)/len(all_gauges)*100:.0f}%) ← no detectable signal')
print(f'  Disagree across datasets:         {len(disagree)} basins  '
      f'({len(disagree)/len(all_gauges)*100:.0f}%) ← dataset-sensitive')

print(f'\n  Basins significant in all 3 (robust glacier signal):')
df_robust = PER_BASIN_A['CR2MET'][
    PER_BASIN_A['CR2MET']['gauge_id'].isin(agree_all)
].sort_values('lat', ascending=False)
print(f'  {"gauge_id":10s} {"name":35s} {"lat":6s} {"cluster"}')
print('  ' + '-' * 65)
for _, row in df_robust.iterrows():
    print(f'  {row["gauge_id"]:10s} {row["gauge_name"]:35s} '
          f'{row["lat"]:6.2f} {row["cluster"]}')
```

## 8. Noah's question — gauged basins with ≤100 km² glacier area

Noah asked whether there are gauged basins with small (≤100 km²) glacier areas
that span a useful latitudinal range. The motivation: very large glacierized basins
(like Baker with >1000 km²) are dominated by glacier dynamics and may not be
ideal for detecting the nuanced relationship between glacier melt and streamflow
in mixed catchments. Smaller glacierized basins where glaciers contribute meaningfully
but don't dominate are the interesting test cases.


```python
# Answer Noah's question
small_glac = []
for gauge_id, info in GAUGE_INFO_FRAC.items():
    if info['glacier_area_km2'] <= 100:
        # Get mean annual fraction from RESULTS
        res = RESULTS['CR2MET'].get(gauge_id)
        mean_frac = np.nan
        if res is not None:
            m  = res['monthly']
            cy = res['complete_years']
            ann = (m[m['year'].isin(cy)]
                   .groupby('year').agg(
                       glac=('glacier_runoff_m3','sum'),
                       q   =('q_m3_mon','sum')))
            ann = ann[ann['q'] > 0]
            if len(ann):
                mean_frac = (ann['glac']/ann['q']).mean() * 100

        small_glac.append({
            'gauge_id':          gauge_id,
            'gauge_name':        info['gauge_name'][:40],
            'lat':               info['cen_lat'],
            'cluster':           info['clusters'][0],
            'glacier_area_km2':  info['glacier_area_km2'],
            'basin_area_km2':    info['basin_area_km2'],
            'glac_pct_basin':    info['glacier_area_km2'] / info['basin_area_km2'] * 100,
            'mean_ann_frac_pct': mean_frac,
        })

noah_df = pd.DataFrame(small_glac).sort_values('lat', ascending=False)
print(f'Gauged basins with ≤100 km² glacier area: {len(noah_df)}')
print(f'Latitude range: {noah_df["lat"].min():.1f}°S to {noah_df["lat"].max():.1f}°S')
print(f'Clusters represented: {sorted(noah_df["cluster"].unique())}')
print()
print(noah_df[['gauge_id','gauge_name','lat','cluster',
               'glacier_area_km2','basin_area_km2',
               'glac_pct_basin','mean_ann_frac_pct']]
      .to_string(index=False, float_format='{:.1f}'.format))

noah_df.to_csv(f'{OUT_DIR}/noah_small_glacierized_basins.csv', index=False)

# Figure: scatter of glacier area vs mean annual fraction, coloured by latitude
fig, ax = plt.subplots(figsize=(10, 6))
sc = ax.scatter(
    noah_df['glacier_area_km2'],
    noah_df['mean_ann_frac_pct'],
    c=noah_df['lat'], cmap='viridis_r',
    s=60, edgecolors='k', lw=0.4, alpha=0.85)
for _, row in noah_df[noah_df['glacier_area_km2'] > 20].iterrows():
    ax.annotate(row['gauge_id'],
                (row['glacier_area_km2'], row['mean_ann_frac_pct']),
                fontsize=6, alpha=0.7,
                xytext=(3, 3), textcoords='offset points')
ax.set_xlabel('Glacier area (km²)', fontsize=11)
ax.set_ylabel('Mean annual glacier fraction (%, CR2MET)', fontsize=11)
ax.set_title(f'Gauged basins with ≤100 km² glacier area (n={len(noah_df)})\n'
             f'Spanning {noah_df["lat"].max():.0f}°S – {noah_df["lat"].min():.0f}°S',
             fontsize=11)
ax.grid(True, alpha=0.3)
plt.colorbar(sc, label='Latitude (°S)')
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/noah_small_glacierized_basins.png',
            dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── WA1 significance diagnostic: extended with distance + ice volume ───────
#
# Compares significant vs non-significant WA1 basins across:
# basin area, glacier area, % glacierized, ice volume, glacier-to-gauge
# distance, mean annual/seasonal precip, temp, and streamflow

from math import radians, sin, cos, sqrt, atan2

def haversine_km(lon1, lat1, lon2, lat2):
    """Straight-line surface distance between two lat/lon points in km."""
    R = 6371.0
    lon1, lat1, lon2, lat2 = map(radians, [lon1, lat1, lon2, lat2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    return R * 2 * atan2(sqrt(a), sqrt(1-a))

# ── Load gauge coordinates from catchment attributes ──────────────────────
attrs = pd.read_csv(f'{CAMELS_FULL}/catchment_attributes.csv',
                    dtype={'gauge_id': str})
gauge_coords = attrs[['gauge_id','gauge_lat','gauge_lon',
                       'outlet_camels_lat','outlet_camels_lon',
                       'outlet_camels_elev','mean_elev',
                       'baseflow_index_1979_2010',
                       'frac_snow_cr2met_1979_2010']].copy()

# ── Load glacier statistics for ice volume ────────────────────────────────
glac_stats_path = (f'{OUTPUT_ROOT}/CR2MET/WA1/RGI_WA1/b_080/L3/summary/'
                   f'glacier_statistics__WA1.csv')
glac_stats = pd.read_csv(glac_stats_path)
print(f'Glacier statistics loaded: {len(glac_stats)} glaciers')
print(f'Volume range: {glac_stats["inv_volume_km3"].min():.4f} '
      f'to {glac_stats["inv_volume_km3"].max():.4f} km³')

# ── Load RGI for WA1 centroids ────────────────────────────────────────────
rgi_bna = pd.read_csv(RGI_PATH)
rgi_wa1 = rgi_bna[rgi_bna['Cluster'] == 'WA1'].copy()
rgi_wa1 = rgi_wa1.merge(
    glac_stats[['rgi_id','inv_volume_km3']],
    left_on='RGIId', right_on='rgi_id', how='left')

wa1_gauge_ids = list(wa1_gauges.keys())

# ── Load CAMELS-CL climate ────────────────────────────────────────────────
precip_df = pd.read_csv(f'{CAMELS_FULL}/precip_cr2met_mm_mon.csv')
tmean_df  = pd.read_csv(f'{CAMELS_FULL}/tmean_cr2met_C_mon.csv')
precip_df = precip_df[(precip_df['year'] >= 2000) &
                      (precip_df['year'] <= 2019)]
tmean_df  = tmean_df[(tmean_df['year'] >= 2000) &
                     (tmean_df['year'] <= 2019)]

SUMMER = [12, 1, 2]
WINTER = [6, 7, 8]

# ── Build characteristics per gauge ──────────────────────────────────────
rows = []
for gauge_id in wa1_gauge_ids:
    info = wa1_gauges[gauge_id]
    gid  = str(gauge_id)

    rgi_ids_here = set(info['rgi_ids'])
    rgi_here     = rgi_wa1[rgi_wa1['RGIId'].isin(rgi_ids_here)]

    # Area-weighted glacier centroid and ice volume
    if len(rgi_here) > 0 and rgi_here['Area'].sum() > 0:
        w          = rgi_here['Area'].values
        glac_lon   = np.average(rgi_here['CenLon'].values, weights=w)
        glac_lat   = np.average(rgi_here['CenLat'].values, weights=w)
        ice_vol    = rgi_here['inv_volume_km3'].sum()
        mean_thick = (ice_vol * 1e9 /
                      (rgi_here['Area'].sum() * 1e6))   # m
    else:
        glac_lon = glac_lat = ice_vol = mean_thick = np.nan

    # Gauge coordinates and catchment attributes
    gc = gauge_coords[gauge_coords['gauge_id'] == gid]
    if len(gc) > 0:
        gauge_lat   = gc['gauge_lat'].values[0]
        gauge_lon   = gc['gauge_lon'].values[0]
        outlet_elev = gc['outlet_camels_elev'].values[0]
        mean_elev   = gc['mean_elev'].values[0]
        bfi         = gc['baseflow_index_1979_2010'].values[0]
        frac_snow   = gc['frac_snow_cr2met_1979_2010'].values[0]
    else:
        gauge_lat = gauge_lon = outlet_elev = np.nan
        mean_elev = bfi = frac_snow = np.nan

    # Distance glacier centroid to gauge
    if not any(np.isnan([glac_lon, glac_lat, gauge_lon, gauge_lat])):
        dist_km = haversine_km(glac_lon, glac_lat, gauge_lon, gauge_lat)
    else:
        dist_km = np.nan

    # Elevation difference
    elev_diff = (mean_elev - outlet_elev
                 if not any(np.isnan([mean_elev, outlet_elev]))
                 else np.nan)

    row = {
        'gauge_id':           gauge_id,
        'gauge_name':         info['gauge_name'][:35],
        'lat':                info['cen_lat'],
        'basin_area_km2':     info['basin_area_km2'],
        'glacier_area_km2':   info['glacier_area_km2'],
        'glac_pct':           info['glacier_area_km2'] /
                              info['basin_area_km2'] * 100,
        'n_glaciers':         info['n_glaciers'],
        'ice_vol_km3':        ice_vol,
        'mean_ice_thick_m':   mean_thick,
        'dist_glac_gauge_km': dist_km,
        'outlet_elev_m':      outlet_elev,
        'mean_basin_elev_m':  mean_elev,
        'elev_diff_m':        elev_diff,
        'baseflow_index':     bfi,
        'frac_snow':          frac_snow,
    }

    # CAMELS-CL climate stats
    for df, col_prefix in [(precip_df, 'precip'), (tmean_df, 'tmean')]:
        if gid not in df.columns:
            for period in ['ann', 'summer', 'winter']:
                row[f'{col_prefix}_{period}'] = np.nan
            continue
        for period, months in [('ann',    list(range(1,13))),
                                ('summer', SUMMER),
                                ('winter', WINTER)]:
            if col_prefix == 'precip':
                row[f'{col_prefix}_{period}'] = (
                    df[df['month'].isin(months)]
                    .groupby('year')[gid].sum().mean())
            else:
                row[f'{col_prefix}_{period}'] = (
                    df[df['month'].isin(months)][gid].dropna().mean())

    # Streamflow from RESULTS
    res = RESULTS['CR2MET'].get(gauge_id)
    if res is not None:
        m  = res['monthly']
        cy = res['complete_years']
        mc = m[m['year'].isin(cy)]
        row['q_ann_km3']    = (mc.groupby('year')['q_m3_mon']
                               .sum().mean() / 1e9)
        row['q_summer_km3'] = (mc[mc['month'].isin(SUMMER)]
                               .groupby('year')['q_m3_mon']
                               .sum().mean() / 1e9)
        row['q_winter_km3'] = (mc[mc['month'].isin(WINTER)]
                               .groupby('year')['q_m3_mon']
                               .sum().mean() / 1e9)
    else:
        row['q_ann_km3'] = row['q_summer_km3'] = row['q_winter_km3'] = np.nan

    rows.append(row)

df_wa1_chars = pd.DataFrame(rows)

# ── Merge significance ─────────────────────────────────────────────────────
# CR2MET: merge beta and r2 too
df_cr2_sig = PER_BASIN_A['CR2MET'].copy()
df_cr2_sig['sig_cr2'] = ((df_cr2_sig[lo_col] > 0) |
                          (df_cr2_sig[hi_col] < 0))
df_wa1_chars = df_wa1_chars.merge(
    df_cr2_sig[['gauge_id','sig_cr2', pred,'r2']].rename(
        columns={pred: 'beta_cr2', 'r2': 'r2_cr2'}),
    on='gauge_id', how='left')

# ERA5: significance only
df_era5_sig = PER_BASIN_A['ERA5'].copy()
df_era5_sig['sig_era5'] = ((df_era5_sig[lo_col] > 0) |
                            (df_era5_sig[hi_col] < 0))
df_wa1_chars = df_wa1_chars.merge(
    df_era5_sig[['gauge_id','sig_era5']],
    on='gauge_id', how='left')

# CRU: significance only
df_cru_sig = PER_BASIN_A['CRU'].copy()
df_cru_sig['sig_cru'] = ((df_cru_sig[lo_col] > 0) |
                          (df_cru_sig[hi_col] < 0))
df_wa1_chars = df_wa1_chars.merge(
    df_cru_sig[['gauge_id','sig_cru']],
    on='gauge_id', how='left')

df_wa1_chars['n_sig']   = (df_wa1_chars[['sig_cr2','sig_era5','sig_cru']]
                            .sum(axis=1))
df_wa1_chars['robust']  = df_wa1_chars['n_sig'] == 3
df_wa1_chars['any_sig'] = df_wa1_chars['n_sig'] >= 1

# ── Summary table: sig vs non-sig ─────────────────────────────────────────
print('=' * 70)
print('WA1 BASINS: SIGNIFICANT vs NON-SIGNIFICANT (CR2MET)')
print('=' * 70)

chars = [
    ('basin_area_km2',      'Basin area (km²)'),
    ('glacier_area_km2',    'Glacier area (km²)'),
    ('glac_pct',            'Glacier % of basin'),
    ('n_glaciers',          'N glaciers'),
    ('ice_vol_km3',         'Total ice volume (km³)'),
    ('mean_ice_thick_m',    'Mean ice thickness (m)'),
    ('dist_glac_gauge_km',  'Glacier-to-gauge distance (km)'),
    ('outlet_elev_m',       'Gauge outlet elevation (m)'),
    ('mean_basin_elev_m',   'Mean basin elevation (m)'),
    ('elev_diff_m',         'Elevation diff basin-outlet (m)'),
    ('baseflow_index',      'Baseflow index'),
    ('frac_snow',           'Fraction precip as snow'),
    ('precip_ann',          'Mean annual precip (mm/yr)'),
    ('precip_summer',       'Mean summer precip DJF (mm)'),
    ('precip_winter',       'Mean winter precip JJA (mm)'),
    ('tmean_ann',           'Mean annual temp (°C)'),
    ('tmean_summer',        'Mean summer temp DJF (°C)'),
    ('tmean_winter',        'Mean winter temp JJA (°C)'),
    ('q_ann_km3',           'Mean annual Q (km³/yr)'),
    ('q_summer_km3',        'Mean summer Q DJF (km³/yr)'),
    ('q_winter_km3',        'Mean winter Q JJA (km³/yr)'),
]

sig = df_wa1_chars[df_wa1_chars['sig_cr2'] == True]
ns  = df_wa1_chars[df_wa1_chars['sig_cr2'] == False]

print(f'\n{"Characteristic":38s} {"Sig (n={})".format(len(sig)):15s} '
      f'{"Not sig (n={})".format(len(ns)):15s} {"Diff"}')
print('-' * 80)
for col, label in chars:
    sig_med = sig[col].median()
    ns_med  = ns[col].median()
    if (ns_med != 0 and not np.isnan(sig_med)
            and not np.isnan(ns_med)):
        diff = f'{(sig_med - ns_med)/abs(ns_med)*100:+.0f}%'
    else:
        diff = 'n/a'
    print(f'{label:38s} {sig_med:15.3f} {ns_med:15.3f} {diff}')

# ── Figure ────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 5, figsize=(22, 9))
axes_flat = axes.flatten()

plot_vars = [
    ('basin_area_km2',      'Basin area (km²)',             True),
    ('glacier_area_km2',    'Glacier area (km²)',            True),
    ('glac_pct',            'Glacier % of basin',            False),
    ('ice_vol_km3',         'Ice volume (km³)',              True),
    ('dist_glac_gauge_km',  'Glacier-to-gauge dist (km)',    False),
    ('outlet_elev_m',       'Gauge elevation (m)',           False),
    ('frac_snow',           'Fraction precip as snow',       False),
    ('baseflow_index',      'Baseflow index',                False),
    ('tmean_winter',        'Winter temp JJA (°C)',          False),
    ('q_summer_km3',        'Summer Q DJF (km³/yr)',         True),
]

color_map = {3: '#0044cc', 2: '#3399ff', 1: '#ff9900', 0: '#cc0000'}
label_map = {3: 'Robust (all 3)', 2: 'Partial (2)',
             1: 'Partial (1)',    0: 'None'}

for ax, (col, label, use_log) in zip(axes_flat, plot_vars):
    for n_sig_val in [3, 2, 1, 0]:
        grp  = df_wa1_chars[df_wa1_chars['n_sig'] == n_sig_val]
        vals = grp[col].dropna()
        lats = grp.loc[vals.index, 'lat']
        if len(vals) == 0:
            continue
        ax.scatter(vals, lats,
                   c=color_map[n_sig_val],
                   marker='o' if n_sig_val > 0 else 'x',
                   s=50 if n_sig_val > 0 else 30,
                   alpha=0.85,
                   edgecolors='k' if n_sig_val > 0
                               else color_map[n_sig_val],
                   lw=0.4,
                   label=label_map[n_sig_val],
                   zorder=3)
    ax.set_xlabel(label, fontsize=8)
    ax.set_ylabel('Latitude (°S)', fontsize=8)
    ax.invert_yaxis()
    ax.grid(True, alpha=0.3)
    if use_log:
        ax.set_xscale('log')
    ax.tick_params(labelsize=7)

axes_flat[0].legend(fontsize=7, loc='lower right')
for ax in axes_flat[len(plot_vars):]:
    ax.set_visible(False)

fig.suptitle(
    'WA1: what distinguishes significant from non-significant glacier signal?\n'
    'Dark blue = robust (all 3)  |  Light blue = 2 datasets  |  '
    'Orange = 1 dataset  |  Red X = none',
    fontsize=11, y=1.01)
plt.tight_layout()
plt.savefig(f'{REG_DIR}/WA1_significance_diagnostic_extended.png',
            dpi=150, bbox_inches='tight')
plt.show()

# ── Full sorted table ─────────────────────────────────────────────────────
print('\nFull WA1 extended table sorted by n_sig then glac_pct:')
display_cols = ['gauge_id','gauge_name','lat',
                'basin_area_km2','glacier_area_km2','glac_pct',
                'ice_vol_km3','dist_glac_gauge_km',
                'outlet_elev_m','frac_snow','baseflow_index',
                'precip_ann','tmean_summer','tmean_winter',
                'q_ann_km3','q_summer_km3',
                'beta_cr2','r2_cr2','sig_cr2','sig_era5','sig_cru','n_sig']
print(df_wa1_chars.sort_values(['n_sig','glac_pct'],
                                ascending=[False, False])
      [display_cols].to_string(
          index=False,
          float_format=lambda x: f'{x:.2f}'
                                 if not np.isnan(x) and abs(x) < 10000
                                 else f'{x:.0f}' if not np.isnan(x)
                                 else 'NaN'))
```

## such interesting results! 

Fraction of precipitation as snow: -90%
This is the single most dramatic difference in the whole table. Significant basins have a median snow fraction of 0.005 (essentially zero — almost all precipitation falls as rain) versus 0.053 for non-significant basins. This directly confirms the temperature story — in significant basins there is virtually no snowpack to compete with glacier melt as a summer water source. When snowmelt is absent, the glacier signal stands out clearly. In non-significant basins, snowmelt and glacier melt are both contributing to summer Q and the regression can't separate them.

Ice volume and thickness: no discriminating power
Total ice volume is 64% lower in significant basins (0.27 km³ vs 0.74 km³) and mean ice thickness is essentially identical (+2%). This is consistent with what we found before — having more ice does not make the glacier signal more detectablea in WA1. What matters is the hydrological context of the basin, not the size of the glaciers.

Baseflow index: -11%
Significant basins have slightly lower baseflow index (0.62 vs 0.70). Higher baseflow index means more of streamflow comes from slow groundwater release rather than direct rainfall-runoff. Non-significant basins tend to have higher baseflow, consistent with the volcanic substrate hypothesis — volcanic aquifers store and slowly release meltwater, creating a steady background flow that buffers the glacier signal and makes it harder to detect statistically.

The emerging WA1 story is now quite clear:
The glacier signal is detectable not where glaciers are largest but where the hydrological context allows it to be distinguished from other sources. Specifically, the detectable basins are warm, low-elevation, rain-dominated systems where: (1) almost no precipitation falls as snow so there is no competing snowmelt signal, (2) gauges are positioned far downstream at low elevations below the snowline, and (3) baseflow from volcanic aquifers is relatively low so the glacier signal isn't buried in a flat groundwater-dominated hydrograph.
The highly glacierized volcanic basins — Allipen, Trancura, Cautin, Lonquimay — fail to show significance not because glaciers don't contribute to their streamflow (we know from the fraction analysis they do) but because their streamflow is dominated by a combination of snowmelt in spring and volcanic groundwater year-round, both of which mask the glacier signal in the regression. The 5.5-month median lag for WA1 is consistent with this — meltwater is being stored in volcanic aquifers and released slowly, appearing in streamflow months later when it's indistinguishable from background flow.
This is a novel finding: in rain-dominated, low-elevation WA1 basins, glacier melt is a statistically detectable driver of streamflow variability precisely because competing water sources are absent — not because glaciers are large. The most glacierized basins in the region paradoxically show the weakest statistical glacier signal due to snowmelt and volcanic groundwater masking.

```python
# ── All-Chile significance diagnostic ─────────────────────────────────────
#
# Replicates the WA1 extended analysis for all 143 basins
# then breaks down by cluster to see regional patterns

from math import radians, sin, cos, sqrt, atan2

def haversine_km(lon1, lat1, lon2, lat2):
    R = 6371.0
    lon1, lat1, lon2, lat2 = map(radians, [lon1, lat1, lon2, lat2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a    = sin(dlat/2)**2 + cos(lat1)*cos(lat2)*sin(dlon/2)**2
    return R * 2 * atan2(sqrt(a), sqrt(1-a))

# ── Load supporting data ───────────────────────────────────────────────────
attrs = pd.read_csv(f'{CAMELS_FULL}/catchment_attributes.csv',
                    dtype={'gauge_id': str})
gauge_coords = attrs[['gauge_id','gauge_lat','gauge_lon',
                       'outlet_camels_elev','mean_elev',
                       'baseflow_index_1979_2010',
                       'frac_snow_cr2met_1979_2010']].copy()

precip_df = pd.read_csv(f'{CAMELS_FULL}/precip_cr2met_mm_mon.csv')
tmean_df  = pd.read_csv(f'{CAMELS_FULL}/tmean_cr2met_C_mon.csv')
precip_df = precip_df[(precip_df['year'] >= 2000) &
                      (precip_df['year'] <= 2019)]
tmean_df  = tmean_df[(tmean_df['year'] >= 2000) &
                     (tmean_df['year'] <= 2019)]

SUMMER = [12, 1, 2]
WINTER = [6, 7, 8]

# Load glacier stats for all clusters
glac_stats_all = []
for cl in CLUSTERS:
    path = (f'{OUTPUT_ROOT}/CR2MET/{cl}/RGI_{cl}/b_080/L3/summary/'
            f'glacier_statistics__{cl}.csv')
    if os.path.exists(path):
        df_cl = pd.read_csv(path)
        df_cl['cluster'] = cl
        glac_stats_all.append(df_cl)
    else:
        print(f'Missing glacier stats: {cl}')

glac_stats = pd.concat(glac_stats_all, ignore_index=True)
print(f'Glacier statistics loaded: {len(glac_stats)} glaciers '
      f'across {len(glac_stats_all)} clusters')

# Load full RGI
rgi_bna = pd.read_csv(RGI_PATH)
rgi_all = rgi_bna.merge(
    glac_stats[['rgi_id','inv_volume_km3']],
    left_on='RGIId', right_on='rgi_id', how='left')

# ── Build characteristics for all 143 basins ─────────────────────────────
rows = []
for gauge_id, info in GAUGE_INFO_FRAC.items():
    gid      = str(gauge_id)
    rgi_ids_here = set(info['rgi_ids'])
    rgi_here     = rgi_all[rgi_all['RGIId'].isin(rgi_ids_here)]

    # Area-weighted glacier centroid and ice volume
    if len(rgi_here) > 0 and rgi_here['Area'].sum() > 0:
        w         = rgi_here['Area'].values
        glac_lon  = np.average(rgi_here['CenLon'].values, weights=w)
        glac_lat  = np.average(rgi_here['CenLat'].values, weights=w)
        ice_vol   = rgi_here['inv_volume_km3'].sum()
    else:
        glac_lon = glac_lat = ice_vol = np.nan

    # Gauge coordinates
    gc = gauge_coords[gauge_coords['gauge_id'] == gid]
    if len(gc) > 0:
        gauge_lat   = gc['gauge_lat'].values[0]
        gauge_lon   = gc['gauge_lon'].values[0]
        outlet_elev = gc['outlet_camels_elev'].values[0]
        mean_elev   = gc['mean_elev'].values[0]
        bfi         = gc['baseflow_index_1979_2010'].values[0]
        frac_snow   = gc['frac_snow_cr2met_1979_2010'].values[0]
    else:
        gauge_lat = gauge_lon = outlet_elev = np.nan
        mean_elev = bfi = frac_snow = np.nan

    # Distance
    if not any(np.isnan([glac_lon, glac_lat, gauge_lon, gauge_lat])):
        dist_km = haversine_km(glac_lon, glac_lat, gauge_lon, gauge_lat)
    else:
        dist_km = np.nan

    row = {
        'gauge_id':           gauge_id,
        'gauge_name':         info['gauge_name'][:35],
        'lat':                info['cen_lat'],
        'cluster':            info['clusters'][0],
        'basin_area_km2':     info['basin_area_km2'],
        'glacier_area_km2':   info['glacier_area_km2'],
        'glac_pct':           info['glacier_area_km2'] /
                              info['basin_area_km2'] * 100,
        'n_glaciers':         info['n_glaciers'],
        'ice_vol_km3':        ice_vol,
        'dist_glac_gauge_km': dist_km,
        'outlet_elev_m':      outlet_elev,
        'mean_basin_elev_m':  mean_elev,
        'baseflow_index':     bfi,
        'frac_snow':          frac_snow,
    }

    # Climate
    for df, col_prefix in [(precip_df,'precip'),(tmean_df,'tmean')]:
        if gid not in df.columns:
            for p in ['ann','summer','winter']:
                row[f'{col_prefix}_{p}'] = np.nan
            continue
        for period, months in [('ann',    list(range(1,13))),
                                ('summer', SUMMER),
                                ('winter', WINTER)]:
            if col_prefix == 'precip':
                row[f'{col_prefix}_{period}'] = (
                    df[df['month'].isin(months)]
                    .groupby('year')[gid].sum().mean())
            else:
                row[f'{col_prefix}_{period}'] = (
                    df[df['month'].isin(months)][gid].dropna().mean())

    # Streamflow
    res = RESULTS['CR2MET'].get(gauge_id)
    if res is not None:
        m  = res['monthly']
        cy = res['complete_years']
        mc = m[m['year'].isin(cy)]
        row['q_ann_km3']    = (mc.groupby('year')['q_m3_mon']
                               .sum().mean() / 1e9)
        row['q_summer_km3'] = (mc[mc['month'].isin(SUMMER)]
                               .groupby('year')['q_m3_mon']
                               .sum().mean() / 1e9)
        row['q_winter_km3'] = (mc[mc['month'].isin(WINTER)]
                               .groupby('year')['q_m3_mon']
                               .sum().mean() / 1e9)
    else:
        row['q_ann_km3'] = row['q_summer_km3'] = row['q_winter_km3'] = np.nan

    rows.append(row)

df_all_chars = pd.DataFrame(rows)

# ── Merge significance ─────────────────────────────────────────────────────
df_cr2_sig = PER_BASIN_A['CR2MET'].copy()
df_cr2_sig['sig_cr2'] = ((df_cr2_sig[lo_col] > 0) |
                          (df_cr2_sig[hi_col] < 0))
df_all_chars = df_all_chars.merge(
    df_cr2_sig[['gauge_id','sig_cr2',pred,'r2']].rename(
        columns={pred:'beta_cr2','r2':'r2_cr2'}),
    on='gauge_id', how='left')

for ds, col in [('ERA5','sig_era5'),('CRU','sig_cru')]:
    df_ds = PER_BASIN_A[ds].copy()
    df_ds[col] = (df_ds[lo_col] > 0) | (df_ds[hi_col] < 0)
    df_all_chars = df_all_chars.merge(
        df_ds[['gauge_id',col]], on='gauge_id', how='left')

df_all_chars['n_sig']   = (df_all_chars[['sig_cr2','sig_era5','sig_cru']]
                            .sum(axis=1))
df_all_chars['robust']  = df_all_chars['n_sig'] == 3
df_all_chars['any_sig'] = df_all_chars['n_sig'] >= 1

# ── Summary table: all Chile ───────────────────────────────────────────────
chars = [
    ('basin_area_km2',      'Basin area (km²)'),
    ('glacier_area_km2',    'Glacier area (km²)'),
    ('glac_pct',            'Glacier % of basin'),
    ('ice_vol_km3',         'Total ice volume (km³)'),
    ('dist_glac_gauge_km',  'Glacier-to-gauge distance (km)'),
    ('outlet_elev_m',       'Gauge outlet elevation (m)'),
    ('mean_basin_elev_m',   'Mean basin elevation (m)'),
    ('baseflow_index',      'Baseflow index'),
    ('frac_snow',           'Fraction precip as snow'),
    ('precip_ann',          'Mean annual precip (mm/yr)'),
    ('precip_summer',       'Mean summer precip DJF (mm)'),
    ('tmean_ann',           'Mean annual temp (°C)'),
    ('tmean_summer',        'Mean summer temp DJF (°C)'),
    ('tmean_winter',        'Mean winter temp JJA (°C)'),
    ('q_ann_km3',           'Mean annual Q (km³/yr)'),
    ('q_summer_km3',        'Mean summer Q DJF (km³/yr)'),
    ('q_winter_km3',        'Mean winter Q JJA (km³/yr)'),
]

print('=' * 75)
print('ALL CHILE: SIGNIFICANT vs NON-SIGNIFICANT (CR2MET)')
print('=' * 75)
sig = df_all_chars[df_all_chars['sig_cr2'] == True]
ns  = df_all_chars[df_all_chars['sig_cr2'] == False]
print(f'\n{"Characteristic":38s} {"Sig (n={})".format(len(sig)):15s} '
      f'{"Not sig (n={})".format(len(ns)):15s} {"Diff"}')
print('-' * 80)
for col, label in chars:
    sig_med = sig[col].median()
    ns_med  = ns[col].median()
    if (ns_med != 0 and not np.isnan(sig_med)
            and not np.isnan(ns_med)):
        diff = f'{(sig_med-ns_med)/abs(ns_med)*100:+.0f}%'
    else:
        diff = 'n/a'
    print(f'{label:38s} {sig_med:15.3f} {ns_med:15.3f} {diff}')

# ── Summary by cluster ─────────────────────────────────────────────────────
print('\n\n' + '=' * 75)
print('BY CLUSTER: median characteristics for significant vs non-significant')
print('=' * 75)

key_chars = ['glacier_area_km2','glac_pct','ice_vol_km3',
             'frac_snow','outlet_elev_m','baseflow_index',
             'tmean_winter','q_summer_km3']

for cl in CLUSTERS:
    df_cl = df_all_chars[df_all_chars['cluster'] == cl]
    if len(df_cl) == 0:
        continue
    sig_cl = df_cl[df_cl['sig_cr2'] == True]
    ns_cl  = df_cl[df_cl['sig_cr2'] == False]
    n_sig  = len(sig_cl)
    n_tot  = len(df_cl)
    print(f'\n{cl} ({n_sig}/{n_tot} significant):')
    print(f'  {"Variable":30s} {"Sig":10s} {"Not sig":10s} {"Diff"}')
    for col in key_chars:
        s = sig_cl[col].median()
        n = ns_cl[col].median()
        if len(sig_cl) == 0 or np.isnan(s) or np.isnan(n) or n == 0:
            continue
        print(f'  {col:30s} {s:10.3f} {n:10.3f} '
              f'{(s-n)/abs(n)*100:+.0f}%')

# ── Figure: all Chile ──────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 5, figsize=(22, 9))
axes_flat = axes.flatten()

plot_vars = [
    ('glacier_area_km2',   'Glacier area (km²)',          True),
    ('glac_pct',           'Glacier % of basin',           False),
    ('ice_vol_km3',        'Ice volume (km³)',             True),
    ('frac_snow',          'Fraction precip as snow',      False),
    ('outlet_elev_m',      'Gauge elevation (m)',          False),
    ('baseflow_index',     'Baseflow index',               False),
    ('tmean_winter',       'Winter temp JJA (°C)',         False),
    ('tmean_summer',       'Summer temp DJF (°C)',         False),
    ('precip_summer',      'Summer precip DJF (mm)',       False),
    ('q_summer_km3',       'Summer Q DJF (km³/yr)',        True),
]

# Color by cluster for non-significant, blue star for significant
cluster_colors = {
    'OT3':'#e6194b','DA1':'#f58231','DA2':'#ffe119',
    'DA3':'#bfef45','WA1':'#3cb44b','WA2':'#42d4f4',
    'WA3':'#4363d8','WA4':'#911eb4','WA5':'#f032e6','WA6':'#a9a9a9'
}

for ax, (col, label, use_log) in zip(axes_flat, plot_vars):
    # Non-significant: colored by cluster
    for cl in CLUSTERS:
        grp = df_all_chars[(df_all_chars['cluster'] == cl) &
                           (df_all_chars['sig_cr2'] == False)]
        vals = grp[col].dropna()
        lats = grp.loc[vals.index, 'lat']
        if len(vals) == 0:
            continue
        ax.scatter(vals, lats, c=cluster_colors[cl],
                   s=25, alpha=0.5, edgecolors='none',
                   label=cl, zorder=2)

    # Significant: gold stars
    sig_grp = df_all_chars[df_all_chars['sig_cr2'] == True]
    sig_vals = sig_grp[col].dropna()
    sig_lats = sig_grp.loc[sig_vals.index, 'lat']
    ax.scatter(sig_vals, sig_lats,
               marker='*', c='gold', s=120,
               edgecolors='k', lw=0.4,
               label='Significant', zorder=5)

    # Robust (all 3): larger gold star with blue edge
    rob_grp  = df_all_chars[df_all_chars['robust'] == True]
    rob_vals = rob_grp[col].dropna()
    rob_lats = rob_grp.loc[rob_vals.index, 'lat']
    ax.scatter(rob_vals, rob_lats,
               marker='*', c='gold', s=200,
               edgecolors='#0044cc', lw=1.0,
               label='Robust (all 3)', zorder=6)

    ax.set_xlabel(label, fontsize=8)
    ax.set_ylabel('Latitude (°S)', fontsize=8)
    ax.invert_yaxis()
    ax.grid(True, alpha=0.3)
    if use_log:
        ax.set_xscale('log')
    ax.tick_params(labelsize=7)

# Legend: clusters + significance markers
handles = [plt.scatter([],[], c=cluster_colors[cl], s=25,
                        label=cl) for cl in CLUSTERS]
handles += [
    plt.scatter([],[], marker='*', c='gold', s=80,
                edgecolors='k', lw=0.4, label='Sig (CR2MET)'),
    plt.scatter([],[], marker='*', c='gold', s=120,
                edgecolors='#0044cc', lw=1.0, label='Robust (all 3)'),
]
axes_flat[0].legend(handles=handles, fontsize=6,
                    loc='lower right', ncol=2)

for ax in axes_flat[len(plot_vars):]:
    ax.set_visible(False)

fig.suptitle(
    'All Chile: what distinguishes significant from non-significant basins?\n'
    'Points colored by cluster  |  Gold star = significant (CR2MET)  '
    '|  Blue-edged star = robust (all 3 datasets)',
    fontsize=11, y=1.01)
plt.tight_layout()
plt.savefig(f'{REG_DIR}/all_chile_significance_diagnostic.png',
            dpi=150, bbox_inches='tight')
plt.show()

# Save full table
df_all_chars.to_csv(f'{REG_DIR}/all_chile_basin_characteristics.csv',
                    index=False)
print(f'\nSaved: {REG_DIR}/all_chile_basin_characteristics.csv')
print(f'Total basins: {len(df_all_chars)}')
print(f'Significant (CR2MET): {df_all_chars["sig_cr2"].sum()}')
print(f'Robust (all 3):       {df_all_chars["robust"].sum()}')
```

These results are fascinating and reveal that the drivers of glacier signal detectability vary dramatically by region. Let me work through the all-Chile picture first, then the regional patterns.

**All-Chile: the headline finding**

The most striking result at the national level is what does NOT discriminate significant from non-significant basins: snow fraction (+1%), baseflow index (+2%), and winter temperature (-6%) are essentially identical between the two groups. This is the opposite of what we found in WA1 specifically, and it tells you something important — the WA1 story about snow fraction and warm temperatures is a regional phenomenon, not a national one. What does discriminate nationally: significant basins have lower gauge elevations (-67%), smaller basin areas (-52%), less glacier area (-76%), less ice volume (-80%), less annual precipitation (-35%), and lower annual and summer Q (-43% and -55%). The pattern suggests that nationally, significant basins tend to be smaller, drier, lower-elevation systems where the glacier signal is large relative to total flow — not the large wet Patagonian systems where precipitation dominates.

**By cluster — the regional story:**

**OT3 (Altiplano, 1/3 significant):** Only one significant basin and the sample is too small to draw strong conclusions. The significant basin has much higher summer Q (+274%) despite less glacier area, suggesting it's a relatively water-rich basin where glacier melt stands out above a small precipitation background.

**DA1 (Dry Andes North, 4/22 significant):** Here the pattern flips from WA1 — significant basins actually have MORE glacier area (+38%) and more ice volume (+55%). In the dry Andes, where precipitation is extremely scarce, larger glaciers produce enough melt to be detectable above the noise of near-zero precipitation variability. This makes physical sense: in an arid basin, any water source that varies month-to-month will be detectable, and larger glaciers vary more in absolute terms. Snow fraction and temperature show almost no difference between significant and non-significant basins, confirming that in DA1 it's simply glacier size that matters.

**DA2 (Dry Andes Central, 3/12 significant):** Significant basins have higher glacier % of basin (+80%) but nearly identical total glacier area and ice volume. This suggests that in DA2, what matters is the glacier area relative to basin size, not the absolute amount of ice. Significant basins also have lower summer Q (-80%), meaning glacier melt is a larger fraction of a small summer streamflow, making it easier to detect. Baseflow index is lower for significant basins (-14%), suggesting less groundwater buffering.

**DA3 (Dry Andes South, 4/16 significant):** The most dramatic pattern in the whole dataset. Significant DA3 basins have 98% less glacier area, 91% less glacier %, and 99% less ice volume than non-significant basins. The significant basins in DA3 are essentially tiny-glacierized catchments. The key discriminator appears to be winter temperature: significant basins are 52% warmer in winter (-1.6°C vs -3.2°C), suggesting they're at lower elevations where snow is less dominant. This partially mirrors the WA1 story — warmer conditions mean less snowmelt competition. But the glacier size result is striking: in DA3, it's the least-glacierized basins that show the strongest signal.

**WA1 (Wet Andes North, 5/44 significant):** Consistent with what we already found. Snow fraction is 93% lower in significant basins, outlet elevation 78% lower, winter temperature 164% warmer. This is the clearest and most consistent regional story in the whole dataset. The WA1 driver of significance is unambiguously the thermal and snow regime — warm, rain-dominated, low-elevation basins.

**WA2 (Wet Andes Central, 5/30 significant):** A completely different pattern from WA1. Significant basins have 97% less glacier area and 98% less ice volume — the significant basins are nearly unglacierized. But they have much higher outlet elevations (+382%: 159 m vs 33 m). This is puzzling and worth investigating further. It may reflect that in WA2, some very small high-elevation basins with minimal glacier cover show detectable glacier signals because they have almost no other water sources competing, while the large glacierized basins at low elevations are dominated by precipitation and river routing.

**WA3 (Northern Patagonia, 5/7 significant):** Most basins here are significant (5/7), which makes sense — these are large Patagonian systems where glaciers are genuinely dominant. The significant basins have smaller glacier area (-60%) and less ice volume (-42%) than non-significant, but higher glacier % of basin (+31%). The non-significant basins are the massive NPI/SPI systems (Baker, Pascua) where total Q is so large that even huge glacier contributions are hard to detect statistically above precipitation variability. Outlet elevation is much lower for significant basins (-84%), consistent with the national pattern.

**WA4 (Southern Patagonia, 5/7 significant):** Same story as WA3 but even more extreme — significant basins have 99% less glacier area and 100% less ice volume. The two non-significant basins are the most heavily glacierized ones. In WA4, the largest glacierized basins are non-significant because their Q is enormous and precipitation-dominated, while small basins with minimal glacier cover show detectable signals because the glacier contribution is large relative to their small total flow.

**WA6 (Tierra del Fuego, 2/2 significant):** All significant, sample too small to compare.

**The synthesis — three distinct regional mechanisms:**

1. **Dry Andes (DA1):** Significance driven by glacier SIZE. In arid basins, bigger glaciers produce more melt that is detectable above near-zero precipitation noise. More ice = stronger signal.

2. **Wet Andes transition zone (WA1, DA3):** Significance driven by THERMAL REGIME. Warmer, lower-elevation basins with less snow have no competing snowmelt signal. Temperature and snow fraction are the key discriminators, not glacier size.

3. **Patagonia (WA3, WA4):** Significance driven by RELATIVE GLACIERIZATION vs TOTAL FLOW. The largest most glacierized systems (Baker, Pascua) are non-significant because precipitation dominates total Q. Smaller basins where glacier melt is large relative to total flow show the strongest signals. Outlet elevation is key — low-elevation gauges integrate enough of the basin to capture glacier contributions.

This three-mechanism framework is a major finding for the paper and should go directly into the discussion. It explains why the conventional wisdom — more glaciers equals stronger glacier signal — only holds in the Dry Andes. Everywhere else, the hydrological context of the basin matters more than the size of the glaciers.

ok, it seems like the finding that small-glacier basins can show statistically significant glacier signals is NOT a modeling artifact — it's a real result driven by the fact that statistical significance and physical magnitude are different things. The appropriate framing is: in many Chilean basins, glacier melt is a statistically detectable predictor of monthly streamflow variability even when its absolute contribution is very small, because it provides a consistent seasonal signal that stands out above precipitation noise. The physically important contributions — where glacier melt actually matters for water resources — remain concentrated in the more glacierized basins and during drought years.
```python
# ── Small basin analysis: glacier signal in smallest 10% of watersheds ────
#
# Alexis's hypothesis: smaller basins relative to glacier size will show
# stronger, more detectable glacier signals with lower uncertainty.
# We look at three definitions of "small":
#   1. Smallest 10% by total basin area
#   2. Smallest 10% by glacier area / basin area ratio (most glacierized)
#   3. Basins where glacier area > 5% of basin area (threshold approach)

pred   = 'coef_glacier_runoff_m3'
lo_col = 'ci_lo_glacier_runoff_m3'
hi_col = 'ci_hi_glacier_runoff_m3'

# Build a combined dataframe with basin characteristics + regression results
basin_chars = pd.DataFrame([
    {
        'gauge_id':          gid,
        'gauge_name':        info['gauge_name'][:40],
        'lat':               info['cen_lat'],
        'cluster':           info['clusters'][0],
        'glacier_area_km2':  info['glacier_area_km2'],
        'basin_area_km2':    info['basin_area_km2'],
        'glac_pct_basin':    info['glacier_area_km2'] / info['basin_area_km2'] * 100,
        'n_glaciers':        info['n_glaciers'],
    }
    for gid, info in GAUGE_INFO_FRAC.items()
])

# Merge with CR2MET per-basin regression results
df_cr2 = PER_BASIN_A['CR2MET'].copy()
df_cr2['sig'] = (df_cr2[lo_col] > 0) | (df_cr2[hi_col] < 0)
df_combined   = basin_chars.merge(
    df_cr2[['gauge_id', pred, lo_col, hi_col, 'sig', 'r2', 'n_obs']],
    on='gauge_id', how='inner')

# ── Definition 1: smallest 10% by basin area ─────────────────────────────
p10_area   = df_combined['basin_area_km2'].quantile(0.10)
small_area = df_combined[df_combined['basin_area_km2'] <= p10_area].copy()

# ── Definition 2: most glacierized 10% by glac_pct_basin ─────────────────
p90_pct   = df_combined['glac_pct_basin'].quantile(0.90)
most_glac = df_combined[df_combined['glac_pct_basin'] >= p90_pct].copy()

# ── Definition 3: glacier area > 5% of basin area ─────────────────────────
thresh_glac = df_combined[df_combined['glac_pct_basin'] >= 5.0].copy()

print('=' * 65)
print('ALL 143 BASINS (baseline):')
print(f'  n = {len(df_combined)}')
print(f'  Significant: {df_combined["sig"].sum()} '
      f'({df_combined["sig"].mean()*100:.0f}%)')
print(f'  Median β_glacier:  {df_combined[pred].median():.2f}')
print(f'  Median R²:         {df_combined["r2"].median():.3f}')
print(f'  Median basin area: {df_combined["basin_area_km2"].median():.0f} km²')
print(f'  Median glac%:      {df_combined["glac_pct_basin"].median():.2f}%')

for label, subset in [
    ('SMALLEST 10% BY BASIN AREA (≤{:.0f} km²)'.format(p10_area), small_area),
    ('MOST GLACIERIZED 10% (≥{:.1f}% glacier cover)'.format(p90_pct), most_glac),
    ('GLACIER AREA > 5% OF BASIN', thresh_glac),
]:
    print(f'\n{label}:')
    print(f'  n = {len(subset)}')
    print(f'  Significant: {subset["sig"].sum()} '
          f'({subset["sig"].mean()*100:.0f}%)')
    print(f'  Median β_glacier:  {subset[pred].median():.2f}')
    print(f'  Median R²:         {subset["r2"].median():.3f}')
    print(f'  Median basin area: {subset["basin_area_km2"].median():.0f} km²')
    print(f'  Median glac%:      {subset["glac_pct_basin"].median():.2f}%')
    print(f'  Lat range:         '
          f'{subset["lat"].min():.1f}°S to {subset["lat"].max():.1f}°S')
    print(f'  Clusters:          '
          f'{sorted(subset["cluster"].unique())}')

# ── Figure ────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Panel 1: significance rate vs glacier % of basin (binned)
ax = axes[0]
bins   = [0, 1, 2, 5, 10, 20, 100]
labels = ['<1%', '1-2%', '2-5%', '5-10%', '10-20%', '>20%']
df_combined['glac_pct_bin'] = pd.cut(
    df_combined['glac_pct_basin'], bins=bins, labels=labels)
sig_rate = df_combined.groupby('glac_pct_bin', observed=True)['sig'].agg(
    ['mean', 'count'])
ax.bar(range(len(sig_rate)), sig_rate['mean'] * 100,
       color='#3498db', edgecolor='k', lw=0.5, alpha=0.8)
for i, (rate, count) in enumerate(
        zip(sig_rate['mean'], sig_rate['count'])):
    ax.text(i, rate * 100 + 1, f'n={count}',
            ha='center', fontsize=8)
ax.set_xticks(range(len(sig_rate)))
ax.set_xticklabels(labels, rotation=30, ha='right', fontsize=8)
ax.set_xlabel('Glacier area as % of basin area', fontsize=10)
ax.set_ylabel('% basins with significant glacier signal', fontsize=10)
ax.set_title('Significance rate vs glacierization\n'
             'More glacierized = stronger signal?', fontsize=9)
ax.grid(True, alpha=0.3, axis='y')

# Panel 2: β_glacier vs glac_pct_basin
ax = axes[1]
ax.scatter(
    df_combined['glac_pct_basin'],
    df_combined[pred],
    c=['#0044cc' if s else '#cc0000' for s in df_combined['sig']],
    s=35, alpha=0.7, edgecolors='k', lw=0.3)
ax.scatter(
    small_area['glac_pct_basin'],
    small_area[pred],
    s=80, facecolors='none', edgecolors='gold',
    lw=1.5, zorder=5, label='Smallest 10% by basin area')
ax.axhline(0, color='k', lw=0.7, ls='--')
ax.set_xlabel('Glacier % of basin area', fontsize=10)
ax.set_ylabel('β_glacier (standardised)', fontsize=10)
ax.set_title('β_glacier vs glacierization\n'
             'Gold = smallest 10% by basin area', fontsize=9)
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

# Panel 3: CI width vs glac_pct_basin (uncertainty quantification)
ax = axes[2]
ci_width       = (df_combined[hi_col] - df_combined[lo_col]).values
ci_width_small = (small_area[hi_col] - small_area[lo_col]).values

ax.scatter(
    df_combined['glac_pct_basin'],
    ci_width,
    c=['#0044cc' if s else '#cc0000' for s in df_combined['sig']],
    s=35, alpha=0.7, edgecolors='k', lw=0.3)
ax.scatter(
    small_area['glac_pct_basin'],
    ci_width_small,
    s=80, facecolors='none', edgecolors='gold',
    lw=1.5, zorder=5, label='Smallest 10% by basin area')
ax.set_xlabel('Glacier % of basin area', fontsize=10)
ax.set_ylabel('CI width (hi - lo)', fontsize=10)
ax.set_title('Uncertainty vs glacierization\n'
             'Narrower CI = more certain estimate', fontsize=9)
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

fig.suptitle(
    'Glacier signal strength and uncertainty vs basin glacierization\n'
    'CR2MET Part A  |  Blue = significant  |  Red = not significant  '
    '|  Gold outline = smallest 10% by basin area',
    fontsize=10, y=1.02)
plt.tight_layout()
plt.savefig(f'{REG_DIR}/small_basin_glacier_signal_analysis.png',
            dpi=150, bbox_inches='tight')
plt.show()

# ── List the smallest 10% basins with their results ───────────────────────
print('\nSmallest 10% of watersheds by basin area — full list:')
print(f'{"gauge_id":10s} {"name":38s} {"lat":6s} {"basin_km2":9s} '
      f'{"glac_km2":8s} {"glac%":6s} {"β":8s} {"R²":6s} {"sig"}')
print('-' * 100)
for _, row in small_area.sort_values('lat', ascending=False).iterrows():
    print(f'{row["gauge_id"]:10s} {row["gauge_name"]:38s} '
          f'{row["lat"]:6.2f} {row["basin_area_km2"]:9.0f} '
          f'{row["glacier_area_km2"]:8.1f} {row["glac_pct_basin"]:6.2f}% '
          f'{row[pred]:+8.2f} {row["r2"]:6.3f} '
          f'{"*" if row["sig"] else ""}')
```

## basin-level regression analysis

```python
# ── Rebuild REG_DATA from RESULTS and GAUGE_INFO_FRAC ─────────────────────
# Standalone — doesn't need livneh_experiments to have run first

print('Building regression dataset from RESULTS...')

CAMELS_FULL = ('/Users/milliespencer/Desktop/CR2_OGGM_Paper/'
               'Full_CAMELS_all_basins/CAMELS_CL_v202201')

# Load CAMELS basin-average climate
print('Loading CAMELS climate...')
precip_wide = pd.read_csv(f'{CAMELS_FULL}/precip_cr2met_mon.csv')
tmean_wide  = pd.read_csv(f'{CAMELS_FULL}/tmean_cr2met_mon.csv')
pet_wide    = pd.read_csv(f'{CAMELS_FULL}/pet_hargreaves_mon.csv')
q_wide_reg  = pd.read_csv(f'{CAMELS_FULL}/q_m3s_mon.csv')

for df in [precip_wide, tmean_wide, pet_wide, q_wide_reg]:
    df.columns = [str(c) for c in df.columns]

REG_DATA = {}

for ds_name in ['CR2MET', 'ERA5', 'CRU']:
    rows = []
    for gauge_id, info in GAUGE_INFO_FRAC.items():
        res = RESULTS[ds_name].get(gauge_id)
        if res is None:
            continue
        m  = res['monthly']
        cy = res.get('complete_years', YEARS)
        mc = m[m['year'].isin(cy)].copy()
        if len(mc) == 0:
            continue

        gid_str = str(gauge_id)

        # Get basin climate
        for df, col_name in [(precip_wide, 'precip_mm'),
                              (tmean_wide,  'tmean_C'),
                              (pet_wide,    'pet_mm'),
                              (q_wide_reg,  'q_m3_mon_obs')]:
            if gid_str in df.columns:
                clim = df[['year','month', gid_str]].copy()
                clim = clim.rename(columns={gid_str: col_name})
                mc   = mc.merge(clim, on=['year','month'], how='left')
            else:
                mc[col_name] = np.nan

        # Convert Q m3s to m3/month if needed
        if 'q_m3_mon_obs' in mc.columns:
            days = mc.apply(
                lambda r: pd.Timestamp(int(r['year']),
                                       int(r['month']), 1).days_in_month,
                axis=1)
            mc['q_m3_mon_obs'] = mc['q_m3_mon_obs'] * days * 86400

        # Use modelled Q from RESULTS (already m3/month)
        mc = mc.rename(columns={'q_m3_mon': 'q_m3_mon'})

        # Seasonal harmonics
        mc['sin_month'] = np.sin(2 * np.pi * mc['month'] / 12)
        mc['cos_month'] = np.cos(2 * np.pi * mc['month'] / 12)
        mc['gauge_id']  = gauge_id
        mc['cluster']   = info['clusters'][0]

        rows.append(mc)

    if rows:
        REG_DATA[ds_name] = pd.concat(rows, ignore_index=True)
        print(f'  {ds_name}: {len(REG_DATA[ds_name]):,} obs')

print('REG_DATA built — now run cluster regression cell')
```

### now trying regression analysis w grouped macro basins instead of subbasins:

```python
# ── Cluster-level regression using grouped basins ─────────────────────────
# Run in the same notebook as the grouped basin analysis
# Uses GAUGE_INFO_FRAC_GRP and RESULTS_GRP already in memory

from sklearn.linear_model import LinearRegression
from scipy import stats
import numpy as np
import pandas as pd

EXCLUDE = {'3022001'}

print('Cluster-level panel regressions — grouped basins (CR2MET)')
print('='*65)
print(f'{"Cluster":6s} {"n_obs":7s} {"R²":6s} {"β_glacier":10s} '
      f'{"p_glacier":10s} {"sig":5s} {"n_basins"}')
print('-'*65)

PREDICTORS = ['glacier_runoff_m3', 'precip_mm',
              'tmean_C',            'pet_mm',
              'sin_month',          'cos_month']
MONTHS_ALL = list(range(1, 13))

# Load CAMELS basin-average climate
CAMELS_FULL = ('/Users/milliespencer/Desktop/CR2_OGGM_Paper/'
               'Full_CAMELS_all_basins/CAMELS_CL_v202201')

precip_wide = pd.read_csv(f'{CAMELS_FULL}/precip_cr2met_mon.csv')
tmean_wide  = pd.read_csv(f'{CAMELS_FULL}/tmean_cr2met_mon.csv')
pet_wide    = pd.read_csv(f'{CAMELS_FULL}/pet_hargreaves_mon.csv')
for df in [precip_wide, tmean_wide, pet_wide]:
    df.columns = [str(c) for c in df.columns]

ds_name = 'CR2MET'
cluster_reg_results = []

# Build regression dataframe for grouped basins
rows = []
for gauge_id, info in GAUGE_INFO_FRAC_GRP.items():
    if gauge_id in EXCLUDE:
        continue
    res = RESULTS_GRP[ds_name].get(gauge_id)
    if res is None:
        continue
    m  = res['monthly']
    cy = res.get('complete_years', YEARS)
    mc = m[m['year'].isin(cy)].copy()
    if len(mc) == 0:
        continue

    gid_str = str(gauge_id)

    # Add basin-average climate
    for df, col in [(precip_wide, 'precip_mm'),
                    (tmean_wide,  'tmean_C'),
                    (pet_wide,    'pet_mm')]:
        if gid_str in df.columns:
            clim = df[['year','month',gid_str]].rename(
                columns={gid_str: col})
            mc = mc.merge(clim, on=['year','month'], how='left')
        else:
            mc[col] = np.nan

    mc['sin_month'] = np.sin(2 * np.pi * mc['month'] / 12)
    mc['cos_month'] = np.cos(2 * np.pi * mc['month'] / 12)
    mc['gauge_id']  = gauge_id
    mc['cluster']   = info['clusters'][0] if info['clusters'] else 'unknown'
    rows.append(mc)

if not rows:
    print('No data — check GAUGE_INFO_FRAC_GRP and RESULTS_GRP are loaded')
else:
    reg_df = pd.concat(rows, ignore_index=True)
    reg_df = reg_df.dropna(subset=PREDICTORS + ['q_m3_mon'])
    reg_df = reg_df[reg_df['q_m3_mon'] > 0]
    print(f'Total obs for regression: {len(reg_df):,}')

    # ── National panel regression ──────────────────────────────────────────
    X_nat = reg_df[PREDICTORS].values
    y_nat = reg_df['q_m3_mon'].values
    mod   = LinearRegression().fit(X_nat, y_nat)
    yp    = mod.predict(X_nat)
    r2_nat = 1 - np.sum((y_nat-yp)**2) / np.sum((y_nat-y_nat.mean())**2)
    print(f'\nNational panel R² = {r2_nat:.3f}')
    print(f'β_glacier = {mod.coef_[0]:.2f}')

    # ── Cluster-level regressions ──────────────────────────────────────────
    print(f'\n{"Cluster":6s} {"n_obs":7s} {"R²":6s} '
          f'{"β_glacier":10s} {"p_glacier":10s} {"sig":5s} {"n_basins"}')
    print('-'*65)

    CLUSTERS_ORDER = ['OT3','DA1','DA2','DA3','WA1',
                      'WA2','WA3','WA4','WA5','WA6']

    for cl in CLUSTERS_ORDER:
        cl_df = reg_df[reg_df['cluster'] == cl].copy()
        if len(cl_df) < 30:
            print(f'{cl:6s} {"too few obs (<30)":>30s}')
            continue

        X = cl_df[PREDICTORS].values
        y = cl_df['q_m3_mon'].values

        mod   = LinearRegression().fit(X, y)
        yp    = mod.predict(X)
        ss_res = np.sum((y - yp)**2)
        ss_tot = np.sum((y - y.mean())**2)
        r2     = 1 - ss_res/ss_tot

        n, k   = len(y), X.shape[1]
        mse    = ss_res / (n - k - 1)
        var_b  = mse * np.linalg.inv(X.T @ X).diagonal()
        se     = np.sqrt(var_b)
        t_stat = mod.coef_[0] / se[0]
        p_val  = 2 * stats.t.sf(abs(t_stat), df=n-k-1)
        sig    = ('***' if p_val < 0.001 else
                  '**'  if p_val < 0.01  else
                  '*'   if p_val < 0.05  else 'ns')

        n_basins = reg_df[reg_df['cluster'] == cl]['gauge_id'].nunique()
        cluster_reg_results.append({
            'cluster':   cl,
            'n_obs':     len(cl_df),
            'n_basins':  n_basins,
            'r2':        r2,
            'b_glacier': mod.coef_[0],
            'p_glacier': p_val,
            'sig':       sig,
        })

        print(f'{cl:6s} {len(cl_df):7,d} {r2:6.3f} '
              f'{mod.coef_[0]:10.2f} {p_val:10.4f} '
              f'{sig:5s} {n_basins:5d}')

    df_cluster_reg = pd.DataFrame(cluster_reg_results)
    print(f'\nSaved cluster regression results')
    print(df_cluster_reg[['cluster','n_basins','r2',
                           'b_glacier','p_glacier','sig']].to_string())
```
